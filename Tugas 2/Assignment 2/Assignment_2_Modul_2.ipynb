## 2. Validasi Input - dari I/O menjadi Pure Validator

```python
                                # (NIM: 202410370110017)

def validate_text(value):
    """
    pre : value berupa string (boleh berspasi di tepi)
    post: (True, teks_bersih) jika tidak kosong, selain itu (False, pesan_error)
    """
    cleaned = str(value).strip()
    return (True, cleaned) if cleaned else (False, "Input tidak boleh kosong.")


def validate_number(value, number_type=float, min_value=0):
    """
    pre : value berupa string/angka, number_type adalah int atau float
    post: (True, angka) jika bisa dikonversi dan >= min_value, selain itu (False, pesan_error)
    """
    try:
        number = number_type(str(value).strip())
    except ValueError:
        return (False, f"'{value}' bukan angka {number_type.__name__}.")
    return (True, number) if number >= min_value else (False, f"Nilai minimal {min_value}.")


def validate_choice(value, options):
    """
    pre : options adalah koleksi angka pilihan yang sah
    post: (True, pilihan_int) jika value ada di options, selain itu (False, pesan_error)
    """
    ok, number = validate_number(value, int, 0)
    return (True, number) if ok and number in options else (False, f"Pilih salah satu dari {tuple(options)}.")


# uji: input sama selalu menghasilkan output sama (deterministik)
print(validate_text("  WiFi "), validate_text("   "))
print(validate_number("48.5"), validate_number("abc"), validate_number("-3", int))
print(validate_choice("2", (1, 2, 3)), validate_choice("9", (1, 2, 3)))
```

## 3. Menu (Slicing) - Pure

```python
# (NIM: 202410370110017)

def get_menu(menu, is_logged_in):
    """
    pre : menu adalah tuple 9 item
    post: tuple menu tamu (index 0-2) atau menu peneliti (index 3-8)
    """
    return menu[3:] if is_logged_in else menu[:3]


def format_menu(menu, is_logged_in, username=None):
    """
    pre : menu adalah tuple menu utama
    post: string menu bernomor sesuai status login (tanpa mencetak)
    """
    title = f"MENU PENELITI ({username})" if is_logged_in else "MENU TAMU"
    lines = [f"{number}. {item}" for number, item in enumerate(get_menu(menu, is_logged_in), start=1)]
    return "\n".join([f"===== {title} ====="] + lines)


def display(text):
    """Lapisan SHELL: satu-satunya fungsi yang melakukan output (impure secara sengaja)."""
    print(text)


display(format_menu(MAIN_MENU, False))
display(format_menu(MAIN_MENU, True, "farid17"))
```

## 4. Registrasi & Login

```python
# (NIM: 202410370110017)

import copy

def register_user(accounts, profiles, user_id, password, name, phone, isp):
    """
    pre : accounts (dict ID->password), profiles (dict ID->dict), seluruh field berupa string
    post: (True, akun_baru, profil_baru, pesan) jika valid dan ID belum dipakai,
          selain itu (False, accounts, profiles, pesan_error). Argumen asli TIDAK diubah.
    """
    checks = [validate_text(v) for v in (user_id, password, name, phone, isp)]
    errors = [msg for ok, msg in checks if not ok]
    if errors:
        return (False, accounts, profiles, "Semua field wajib diisi.")
    user_id, password, name, phone, isp = [value for _, value in checks]
    if user_id in accounts:
        return (False, accounts, profiles, f"ID '{user_id}' sudah digunakan.")

    new_accounts = copy.deepcopy(accounts)
    new_profiles = copy.deepcopy(profiles)
    new_accounts[user_id] = password
    new_profiles[user_id] = {"name": name, "phone": phone, "role": "Peneliti", "isp": isp}
    return (True, new_accounts, new_profiles, f"Registrasi '{user_id}' berhasil.")


def login_user(accounts, user_id, password):
    """
    pre : accounts (dict ID->password)
    post: user_id jika kredensial cocok, selain itu None
    """
    return user_id if accounts.get(user_id) == password else None


def greeting(profiles, user_id):
    """post: kalimat sambutan untuk user_id (pure, tanpa print)"""
    profile = profiles[user_id]
    return f"Selamat datang, {profile['name']} ({profile['role']})!"
```

## 5. Manipulasi Data Eksperimen (Session Lock)

```python
# (NIM: 202410370110017)

def get_my_tests(tests, username):
    """
    pre : tests (list of dict), username (str)
    post: list BARU berisi catatan milik username saja (session lock)
    """
    return [test for test in tests if test["owner"] == username]


def find_my_test(tests, username, test_id):
    """
    pre : test_id (int)
    post: dict catatan milik username dengan test_id tsb, atau None
    """
    return next((t for t in get_my_tests(tests, username) if t["test_id"] == test_id), None)


def validate_test_input(location, connection, download, upload, ping, stable):
    """
    pre : nilai mentah (string/angka) hasil formulir
    post: (True, dict_bersih) jika semua valid, selain itu (False, list_pesan_error)
    """
    results = {
        "location": validate_text(location),
        "connection": validate_text(connection),
        "download_mbps": validate_number(download, float),
        "upload_mbps": validate_number(upload, float),
        "ping_ms": validate_number(ping, int),
    }
    errors = [f"{field}: {msg}" for field, (ok, msg) in results.items() if not ok]
    if errors:
        return (False, errors)
    return (True, {**{field: value for field, (_, value) in results.items()}, "stable": bool(stable)})


def add_speedtest(tests, username, location, connection, download, upload, ping, stable):
    """
    pre : tests (list of dict), username sudah login
    post: (True, list_baru_dengan_catatan_tambahan, pesan) atau (False, tests, pesan_error).
          List asli tidak diubah.
    """
    ok, data = validate_test_input(location, connection, download, upload, ping, stable)
    if not ok:
        return (False, tests, "; ".join(data))
    next_id = max([t["test_id"] for t in tests], default=0) + 1
    new_tests = copy.deepcopy(tests)
    new_tests.append({"test_id": next_id, "owner": username, **data})
    return (True, new_tests, f"Hasil speedtest #{next_id} tersimpan.")


def correct_speedtest(tests, username, test_id, download, upload, ping, stable):
    """
    pre : test_id milik username
    post: (True, list_baru_dengan_catatan_terkoreksi, pesan) atau (False, tests, pesan_error)
    """
    if find_my_test(tests, username, test_id) is None:
        return (False, tests, "Catatan tidak ditemukan atau bukan milik Anda.")
    checks = {"download_mbps": validate_number(download, float),
              "upload_mbps": validate_number(upload, float),
              "ping_ms": validate_number(ping, int)}
    errors = [f"{f}: {m}" for f, (ok, m) in checks.items() if not ok]
    if errors:
        return (False, tests, "; ".join(errors))
    changes = {**{f: v for f, (_, v) in checks.items()}, "stable": bool(stable)}
    source = copy.deepcopy(tests)
    new_tests = [{**t, **changes} if t["test_id"] == test_id else t for t in source]
    return (True, new_tests, f"Catatan #{test_id} berhasil dikoreksi.")


def discard_speedtest(tests, username, test_id):
    """
    pre : test_id milik username
    post: (True, list_baru_tanpa_catatan_tsb, pesan) atau (False, tests, pesan_error)
    """
    if find_my_test(tests, username, test_id) is None:
        return (False, tests, "Catatan tidak ditemukan atau bukan milik Anda.")
    return (True, [copy.deepcopy(t) for t in tests if t["test_id"] != test_id], f"Catatan #{test_id} dihapus.")
```

```python
# (NIM: 202410370110017)

def format_history(tests, username):
    """
    post: string riwayat speedtest milik username (tanpa print)
    """
    my_tests = get_my_tests(tests, username)
    if not my_tests:
        return f"--- Riwayat {username} ---\n  Belum ada catatan eksperimen."
    rows = [
        f"  #{t['test_id']} | {t['location']} ({t['connection']}) | DL {t['download_mbps']} Mbps | "
        f"UL {t['upload_mbps']} Mbps | Ping {t['ping_ms']} ms | {'Stabil' if t['stable'] else 'Tidak stabil'}"
        for t in my_tests
    ]
    return "\n".join([f"--- Riwayat {username} ---"] + rows)


def average(values):
    """pre: list angka | post: rata-rata, atau 0 jika list kosong"""
    return sum(values) / len(values) if values else 0


def summarize_speed(tests, username):
    """
    pre : tests (list of dict), username (str)
    post: dict ringkasan statistik milik username, atau None jika belum ada data
    """
    my_tests = get_my_tests(tests, username)
    if not my_tests:
        return None
    fastest = max(my_tests, key=lambda t: t["download_mbps"])
    return {
        "total": len(my_tests),
        "avg_download": average([t["download_mbps"] for t in my_tests]),
        "avg_upload": average([t["upload_mbps"] for t in my_tests]),
        "avg_ping": average([t["ping_ms"] for t in my_tests]),
        "stable_ratio": average([1 if t["stable"] else 0 for t in my_tests]) * 100,
        "fastest": (fastest["location"], fastest["download_mbps"]),
    }


def format_summary(summary, username):
    """post: string ringkasan dari dict hasil summarize_speed()"""
    if summary is None:
        return f"--- Ringkasan {username} ---\n  Belum ada data untuk diringkas."
    return "\n".join([
        f"--- Ringkasan Kecepatan {username} ---",
        f"  Jumlah pengujian  : {summary['total']}",
        f"  Rata-rata download: {summary['avg_download']:.2f} Mbps",
        f"  Rata-rata upload  : {summary['avg_upload']:.2f} Mbps",
        f"  Rata-rata ping    : {summary['avg_ping']:.1f} ms",
        f"  Koneksi stabil    : {summary['stable_ratio']:.0f}%",
        f"  Lokasi tercepat   : {summary['fastest'][0]} ({summary['fastest'][1]} Mbps)",
    ])
```
