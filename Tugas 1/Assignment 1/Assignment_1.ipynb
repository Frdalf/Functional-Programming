## 1. Function 1 & Function 2 (dual output)

```python
# (NIM: 202410370110017)
def calculate_final_scores(grades):
    #Function 1: hitung Final Score, kembalikan list BARU berisi 4 elemen.
    #Data asli (grades) tidak diubah -> tidak ada efek samping.
    scored = []
    for student in grades:
        name, assignment, exam = student
        final = round((assignment * 0.4) + (exam * 0.6), 2)
        scored.append([name, assignment, exam, final])
    return scored


def filter_students(scored_grades):
    #Function 2: pisahkan nama mahasiswa lulus & tidak lulus.
    #Syarat lulus: Exam >= 60 AND Final >= 70.
    passed = []
    failed = []
    for name, assignment, exam, final in scored_grades:
        if exam >= 60 and final >= 70:
            passed.append(name)
        else:
            failed.append(name)
    return passed, failed  # dual output


scored_grades = calculate_final_scores(raw_grades)
passed_list, failed_list = filter_students(scored_grades)

print("=== Function 1: Data dengan Final Score ===")
for row in scored_grades:
    print(row)

print("\n=== Function 2: Hasil Filter ===")
print("Lulus       :", passed_list)
print("Tidak Lulus :", failed_list)
```

## 2. Konversi ke Tuple (NIM ganjil -> daftar LULUS)

```python
# (NIM: 202410370110017)
passed_tuple = tuple(passed_list)

print("\n=== Konversi ke Tuple ===")
print("Tuple     :", passed_tuple)
print("Tipe data :", type(passed_tuple))
```

## 3. Rekursi + Exception Handling

```python
# (NIM: 202410370110017)
def count_students_recursive(students, index=0):
    #Hitung jumlah elemen tanpa for/while/len().
    #Base case: slicing dari index menghasilkan tuple kosong.
    if students[index:] == ():
        return 0
    return 1 + count_students_recursive(students, index + 1)


def verify_student(name, students):
    #Verifikasi nama dengan try-except-else.
    #Metode tuple.index() otomatis melempar ValueError jika nama tidak ada.
    try:
        position = students.index(name)
    except ValueError:
        print(f"[ERROR] '{name}' tidak ditemukan dalam daftar mahasiswa lulus.")
    else:
        print(f"[OK] '{name}' terverifikasi lulus (urutan ke-{position + 1}).")


print("\n=== Rekursi ===")
print("Jumlah mahasiswa lulus:", count_students_recursive(passed_tuple))

print("\n=== Verifikasi Mahasiswa ===")
verify_student("Amel", passed_tuple)
verify_student("Rani", passed_tuple)
```
