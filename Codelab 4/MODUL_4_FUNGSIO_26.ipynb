# 1. LAMBDA EXPRESSION

In programming, not every function we create has complex logic or is intended for repeated use. Sometimes, we only need a simple function to perform a single operation, such as multiplying a number, formatting data, or checking a specific condition. Writing a complete function using the def keyword for these simple tasks can often feel unnecessarily verbose compared to the actual logic we want to execute.

To address this, Python provides **lambda expressions** as a concise way to define simple functions in a single line. A lambda expression does not require an explicit function name, so **it is commonly referred to as an anonymous function**.

In mathematics, a function is commonly written as:


$$f(x)=2x$$



````python
def f(x):
    return x*2
````

````python
print("Output of Standard Function (def): ", f(10))
````

The function above accepts a single parameter `x `and returns the result of multiplying `x` by `2`. The same operation can be expressed using a lambda expression:

````python
f_lambda  = lambda x : x*2
````

The lambda expression lambda `x: 2 * x` represents a function that takes `x `as its parameter and returns `2 * x`. Unlike a function defined using def, the lambda expression does not explicitly provide a function name.

The general syntax of a lambda expression is:
>`lambda parameter: expression`

In this syntax, parameter represents the input received by the function, while expression represents the operation whose result is automatically returned. Therefore, in lambda `x: 2 * x`, `x `**is the parameter** and `2 * x `**is the expression**. A lambda expression does not require an explicit return statement.


Although more concise, a lambda expression behaves the same way as a function defined using def. Both accept an input value and produce an output value. **The main difference is in how they are defined and written.**

````python
print("the type of 'f_lambda':", type(f_lambda))
print("Output of Lambda Expression:", f_lambda(10))
````

The output shows that `f_lambda` is a **function object**, even though it was defined using a lambda expression rather than `def`. This means that a lambda expression can be assigned to a variable, passed as an argument to another function, or called like a regular function.

Because of their concise syntax, lambda expressions are particularly useful when a simple function is needed for a specific operation, especially when the function does not require a separate name or a complex implementation.


## 1.1 Assigned to a Variable

A lambda expression can be assigned to a variable, allowing the resulting function to be called using that variable. This makes it possible to use a lambda function in a similar way to a regular function, while keeping its definition concise.

Consider the following example.

````python
# lambda to convert text to uppercase
uppercase = lambda text: text.upper()
print(uppercase("informatics"))

# lambda to calculate the area of a rectangle
rectangle_area = lambda length, width: length * width
print(rectangle_area(8, 5))
````

The first lambda function is assigned to the variable `uppercase` to convert text to uppercase using the `upper()` method, while the second lambda function is assigned to the variable `rectangle_area` to calculate the area of a rectangle based on its `length` and `width`. Both lambda functions are then invoked just like regular functions by calling the variables to which they were assigned.

## 1.2 Passed as a Function Argument

In addition to being assigned to a variable, a lambda function can also be passed directly as an argument to another function. This is one of the main advantages of lambda expressions because there is no need to define a separate function using def when the function is intended for one-time use. This technique is commonly used with Python's built-in functions such as **map(), filter(),** and **reduce()**, which were introduced in the previous module.

### 1.2.1 MAP

The **map(function, iterable)** function applies a given function to each element of an iterable (such as a list or tuple).

Using a lambda function allows the transformation logic to be written directly without first defining a separate function with def.

````python
celsius = [0, 10, 20, 30, 40]
````

````python
# Using map() to convert temperatures from Celsius to Fahrenheit
fahrenheit = map(lambda temp: (temp * 9/5) + 32, celsius)

print(list(fahrenheit))
````

In the example above, the lambda function is passed as the first argument to `map()`. Each value in the celsius list is passed one by one to the parameter temp, where it is converted from Celsius to Fahrenheit using the formula defined in the lambda expression. The transformed values are then returned by `map()` as an iterator and converted into a list so that they can be displayed.

This example also demonstrates the concepts of First-Class Functions and Higher-Order Functions (HOFs). The lambda expression is treated as a function object that can be passed as an argument, while `map()` is considered a Higher-Order Function because it accepts another function as one of its arguments.

If the same operation needs to be performed multiple times, the lambda function can first be assigned to a variable to improve code readability and reusability. Like this one:

````python
to_fahrenheit = lambda temp: (temp * 9/5) + 32

fahrenheit = map(to_fahrenheit, celsius)
print(list(fahrenheit))
````

In the second example, the lambda function is assigned to the variable `to_fahrenheit`, which is then passed to `map()`. This approach produces the same output as the previous example, but makes the function easier to reuse in multiple parts of a program.

### 1.2.2 FILTER

The **filter(function, iterable)** function is used to select elements from an iterable based on a specified condition. Each element is evaluated by the given function. If the function returns True, the element is included in the result otherwise, it is excluded.

A lambda function is commonly used with `filter()` because it allows the filtering condition to be written directly without defining a separate function using def.

````python
scores = [35, 67, 82, 45, 91, 58]
````

````python
# Using filter() to select scores greater than or equal to 60
passed = filter(lambda score: score >= 60, scores)

print(list(passed))
````

In the example above, the `lambda` function is passed as the first **argument** to `filter()`. Each value in the scores list is **evaluated** against the condition `score >= 60`. Only the values that satisfy this condition are retained in the final result.


Another example of using `filter()` to filter text data:

````python
words = ("computer", "pen", "notebook", "book", "keyboard")
````

````python
# Using filter() to select words with more than five characters
long_words = filter(lambda word: len(word) > 5, words)

print(tuple(long_words))
````

Just like the previous example, the lambda function only returns True for words with more than five characters. As a result, only those words are included in the final result.

>Note that `filter()` uses lazy evaluation and returns an iterator rather than the final values. Because of this, its output cannot be printed directly and must first be converted into a data structure like a list or tuple.

## 1.3 Direct Execution


Besides being assigned to a variable or passed as an argument to another function, a lambda function can also be executed immediately after it is defined. This approach is useful for simple operations that are performed only once and do not need to be reused elsewhere in the program.

````python
# Immediately executing a lambda function
celcius = 100
farenheit = (lambda temp: (temp * 9/5) + 32)(celcius)
print(farenheit)
````

In the example above, the lambda function is defined and executed in a single statement. The value of `celsius` is passed directly to the lambda function as an argument, and the resulting value is assigned to the `fahrenheit` variable. This demonstrates that a lambda function can be executed directly without first assigning the lambda itself to a variable. Functions that work with multiple arguments can also be expressed in other forms in functional programming. One of these techniques is **currying**, which changes how arguments are provided to a function. The concept of currying will be explored in more detail in the following section.


## 1.4 Limitations of Lambda Functions

Lambda functions are great for writing quick, concise operations. However, they have several key limitations you should keep in mind:


1. A lambda function is limited to a single expression. Unlike a regular function defined with def, it cannot contain multiple statements or a block of code.
```
square = lambda x: x ** 2
print(square(5))
```

2. Conditional expressions are limited to a single inline expression. A lambda function may use a simple conditional expression (if ... else), but it cannot contain multiple conditional branches (if-elif-else) or loop statements such as for and while.
```
grade = lambda score: "Pass" if score >= 60 else "Fail"
```

Although lambda functions cannot replace regular functions created with def when implementing complex logic, they are particularly useful for short, one-time operations. When combined with the concept of First-Class Functions, lambda expressions make Python code more concise, expressive, and readable.

Here is an example of a function that cannot be converted directly by lambda:

````python
# Function that assigns a letter grade based on a student's score
def calculate_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'
````

````python
# use the function above
score = 85
grade = calculate_grade(score)

print("Score:", score)
print("Grade:", grade)
````

The function contains multiple `if-elif-else` statements, making it more complex than the simple operations typically intended for lambda functions. Therefore, using a regular function with `def` makes the code easier to read and understand.

>Note: Python technically allows this logic to be written as a lambda using nested conditional expressions. However, as the number of conditions increases, the resulting lambda becomes difficult to read. For this reason, `def` is generally preferred for functions with more complex logic.

# 2. CLOSURE

Now that you can write inline functions using lambdas, you'll often encounter situations where an inner lambda needs to remember variables from outside its body. This capability is called a closure.

A closure happens when an inner function remembers variables from its outer function even after that outer function has finished running.

 To better understand this concept, consider the following example program.

````python
#cell 2a
def create_multiplier(factor): # as outer function
    def multiply(number): # as inner function
        return number * factor
    return multiply
````

or it can be written like this if using a lambda:

````python
#cell 2b
def create_multiplier(factor): # as outer function
    return lambda number: number * factor # as inner function
````

In the example above, `create_multiplier()` acts as the outer function and returns a lambda expression as the inner function. The `factor` variable is a parameter of `create_multiplier()`, while the lambda expression has its own parameter, `number`. The lambda can still access the `factor` variable even though `factor` is not defined directly inside the lambda. This is possible because the lambda expression is defined within the scope of `create_multiplier()` and retains access to the `factor` variable.

Next, run the following code.

````python
#cell 2c
multiplier = create_multiplier(4)
print(multiplier(10))
````

When `create_multiplier(4)` is called, the value `4` is assigned to `factor`. The function then creates and returns the inner lambda function. The returned lambda is stored in `multiplier`.

When `multiplier(10)` is executed, `10` is passed to `number`. Because of closure, multiplier remembers that `factor` was set to `4`, even after `create_multiplier()` has finished running. Therefore, it calculates `10 * 4` and returns `40`.

This demonstrates a **Closure**, where a function retains access to variables from its enclosing scope even after the outer function has finished executing. In this example, the lambda retains access to `factor` from `create_multiplier()`.

## 2.1 CLOSURE VS FIRST CLASS FUNCTION

After learning about First-Class Functions and Closures, you might wonder:

Are Closures the same as First-Class Functions?

**The answer is no**

A First-Class Function is a property of a programming language that allows functions to be treated like any other object. As a result, a function can be assigned to a variable, passed as an argument to another function, or returned as the result of a function.

A Closure, on the other hand, is a function that remembers and can still access variables from its enclosing (outer) scope, even after the outer function has finished executing.

The relationship between these two concepts is that Closures are possible because Python supports First-Class Functions. Since Python allows a function to be returned as a value, the returned function can retain access to the variables that existed when it was created. This ability enables a Closure to "remember" variables from its enclosing scope.

>In other words:
>- First-Class Functions describe what a function can do.
>- Closures describe what a returned function can remember

# 3. CURRYING

You've seen how closures allow a function to retain variables from its outer scope. Now, we can leverage this behavior to write functions in a powerful new way called currying.



## 3.1 What is Currying?

Currying is a technique in functional programming where a function that takes multiple arguments is transformed into a sequence of nested functions that each take a single argument.

Instead of providing all arguments at once, the arguments are supplied one at a time through successive function calls.

In general, a function written as:

$$f(a, b, c)$$



can be transformed into:

$$f(a)(b)(c)$$

This means that the first function accepts the argument (a) and returns another function. The returned function then accepts the argument (b) and produces another function that waits for the next argument (c). This process continues until all required arguments have been provided, at which point the final result is returned.

By transforming a multi-parameter function into a chain of single-parameter functions, currying improves flexibility and allows functions to be reused in different contexts.

Consider the following regular function that calculates the final price after applying a discount.

````python
# Regular function to calculate the final price after a discount
def calculate_discount(price, discount):
    return price - (price * discount / 100)
````

````python
print(calculate_discount(200000, 20))
````

In the example above, both arguments (price and discount) are supplied at the same time, allowing the function to immediately calculate and return the final price.

The same function can be rewritten using currying as follows.

````python
# Curried version of the discount function
def calculate_discount(discount):
    def apply_discount(price):
        return price - (price * discount / 100)
    return apply_discount
````

In this implementation, the parameters are not simply split into separate functions in their original order. It is important to consider **which parameter should be provided first and retained for later use**. Here, `discount` is placed in the outer function because its value is intended to remain fixed, while `price` is placed in the inner function because it can vary for each calculation.



This design is closely related to the concept of a **Closure**. When `calculate_discount(20)` is called, it creates and returns the `apply_discount()` function, which retains access to `discount = 20`. The returned function can then be called with different prices, without providing the discount again. This ability to retain access to the `discount` value is provided by the Closure.
Consider the following example:


````python
#a function that always applies a 20% discount
discount20 = calculate_discount(20)

print(discount20(150000))
print(discount20(300000))
print(discount20(500000))
````

````python
# Or we can call both in a single chained execution:
print(calculate_discount(20)(200000))
````


Unlike the previous example, `calculate_discount(20)(200000)` executes the curried function directly in a single statement. While the closure still occurs behind the scenes, this inline approach evaluates the result immediately rather than saving the intermediate function to a variable.

You can choose which one to use depending on your situation:

- **Save to a variable**: Use this when you want to preset a value (like a fixed 10% discount) **and reuse it multiple times**.

- **Direct chained call `f(x)(y)`**: Use this when you have both values ready at once and **just want a quick, one-line result**.

## 3.2 Lambda and Currying

In addition to using regular functions (`def`), the currying concept can also be implemented using lambda expressions. Lambda expressions provide a more concise way to write the same logic while preserving the behavior and functionality of the original implementation.

For example, the `calculate_discount()` function introduced in Section 3.1 can be rewritten using a lambda expression as follows:

````python
calculate_discount = lambda discount: (
    lambda price: price - (price * discount / 100)
)
````

This implementation uses nested lambda functions. The outer lambda accepts the `discount` parameter and returns an inner lambda that accepts the `price` parameter.

Because each function accepts only a single argument, this directly demonstrates the concept of currying. Furthermore, the inner lambda uses a closure to remember the `discount` value when calculating the final price.

The function can then be called as follows:

````python
# 1. Pass the first argument (discount)
ten_percent_off = calculate_discount(20)  # Returns a curried lambda waiting for price

# 2. Pass the second argument (price)
final_price = ten_percent_off(100)        # Returns 90.0

# Or call both in a single chained execution:
print(calculate_discount(20)(100))        # Output: 90.0
````

When `calculate_discount(20)` is called, it does not immediately calculate the final price. Instead, it returns another lambda expression that stores the 20% discount rate and waits for the `price` argument. Once the `price` is provided in the next function call, the calculation is performed and the final result is returned.

Therefore, implementing currying with lambda expressions produces the same behavior as the nested-function implementation presented in Section 3.1. The only difference lies in the syntax, where lambda expressions provide a more concise way to define simple functions.

## 3.3 Currying and Higher-Order Functions

Although their implementations may appear similar, Currying and Higher-Order Functions (HOFs) are **different concepts**.

- A **Higher-Order Function** is a function that accepts another function as an argument, returns a function as its result, or both.
- In contrast, **Currying** is a functional programming technique that transforms a function with multiple parameters into a sequence of functions, where each function accepts only one parameter.

The two concepts are **closely related** because *Currying is commonly implemented using Higher-Order Functions*. When a function is curried, one function returns another function that waits for the next argument. Therefore, the returned function makes the implementation a Higher-Order Function as well.

However, being a *Higher-Order Function does not automatically mean that a function uses Currying*. A Higher-Order Function only needs to accept or return another function, while Currying specifically focuses on transforming a multi-parameter function into a sequence of single-parameter functions.

In simple terms:

* **Higher-Order Function:** focuses on **functions being passed or returned as values**.
* **Currying:** focuses on **providing function parameters one at a time**.

Understanding this distinction is important because the two concepts *can be used together, but they serve different purposes* in functional programming.


## 3.4 WHY IS CURRYING USEFUL?

Currying is more than just an alternative way of writing functions. It provides several practical benefits, particularly in programs that adopt the functional programming paradigm. Two of the main advantages are function reusability and simpler function composition.

### a. Reusable Functions

One of the greatest advantages of currying is that it allows us to create specialized functions without rewriting the same logic.

Consider the calculate_discount() function introduced in Section 2.1, we can create multiple functions for different discount rates.

````python
discount10 = calculate_discount(10)
discount20 = calculate_discount(20)

print(discount10(150000))
print(discount20(150000))
````

In this example, the calculation logic is written only once inside calculate_discount(). By supplying different values for the first argument, we can create specialized functions such as discount10 and discount20 without duplicating the implementation.

As a result, the code becomes more modular, easier to maintain, and easier to reuse. In addition, currying encourages the development of small functions that perform a single responsibility, making them easier to test and combine with other functions.

### b. Simplifying Function Composition

Currying also makes it easier to combine multiple operations into a sequence of function calls. Since each function accepts only one argument, intermediate functions can be reused before completing the final computation.

For example, consider the following functions:

````python
def add(value):
    return lambda number: number + value

def multiply(value):
    return lambda number: number * value
````

We can create reusable functions and combine them step by step.

````python
# Create a function that always adds 5 to its input
add5 = add(5)

# Create a function that always multiplies its input by 2
multiply2 = multiply(2)

# First, add 5 to 10, then multiply the result by 2
result = multiply2(add5(10))

# Display the final result
print(result)
````

`add5(10)`is evaluated first, producing the value `15`. The result is then passed to `multiply2()`, which multiplies it by `2` and returns `30`.

Because each function accepts a single argument, the processing steps become more modular and easier to understand. This approach also promotes code reuse, as the same functions (`add5` and `multiply2`) can be applied to different input values without modifying their implementations.

## 3.5 CURRYING VS PARTIAL APPLICATION

Currying and Partial Application are two closely related concepts in functional programming. Since both involve supplying function arguments step by step, they are often considered the same concept. However, they have different purposes and should be distinguished.

**Currying**

Currying is a technique that transforms a function with multiple parameters into a sequence of functions, where each function accepts only one parameter.

For example, the following function:

````python
def make_drink(ingredient1):
    def step2(ingredient2):
        def step3(ingredient3):
            return f"Drink made with {ingredient1}, {ingredient2}, and {ingredient3}"
        return step3
    return step2
````

````python
print(make_drink("hot water")("sugar")("tea"))
````

In this example, each function call accepts only one argument, which is the main characteristic of currying.

**Partial Application**

Once a function has been transformed into its curried form, it is not necessary to provide all arguments immediately. Instead, we can supply only some of the required arguments and save the returned function for later use. This technique is known as Partial Application.

Consider the following example.

````python
# Provide the first two ingredients
hot_drinks = make_drink("hot water")("sugar")

# Provide the final ingredient
make_coffee = hot_drinks("coffee")
make_tea = hot_drinks("tea")
make_ginger_drink = hot_drinks("crushed ginger")

print(make_coffee)
print(make_tea)
print(make_ginger_drink)
````

In the example above, the make_drink() function requires three arguments. However, only the first two arguments ("hot water" and "sugar") are provided. As a result, the function does not produce the final output immediately. Instead, it returns another function, which is stored in the hot_drinks variable and waits for the remaining argument.

When only some of the required arguments are supplied while others are left for later, the process is called Partial Application. In other words, Partial Application occurs when a curried function is invoked with only a subset of its arguments, producing a more specialized function.

Since hot_drinks has already fixed the first two arguments, it can be reused with different third ingredients, such as "coffee", "tea", or "crushed ginger", without rewriting the original function.



In short:
- Currying is a technique that transforms a function with multiple parameters into a sequence of functions, where each function accepts only one parameter.
- Partial Application is a technique that supplies only some of the arguments to a curried function, producing a new function that waits for the remaining argument(s)

# 4. DECORATOR

In the previous module, we learned that functions in Python can be treated as objects (First-Class Functions), passed as arguments or returned by other functions (Higher-Order Functions), and even remember variables from their enclosing scope (Closures).

A Decorator is one of the practical applications of these concepts. In simple terms, a Decorator is a function that extends or modifies the behavior of another function without changing its original implementation.

A Decorator has three main characteristics:
1. It accepts another function as an argument.
2. It returns a new function (commonly called a wrapper).
3. It adds or modifies the behavior of the original function.

Consider the following example:

````python
# Decorator
def uppercase_decorator(func):
    def wrapper():
        message = func()
        return message.upper()
    return wrapper
````

Based on these characteristics, we can see that a **decorator is essentially a Higher-Order Function** that is specialized to extend or modify another function's behavior.

````python
# Original function
def greeting():
    return "Welcome to Functional Programming"

print(greeting())
````

The `greeting()` function simply returns a text message. Without modifying the original function, we can extend its behavior by passing it to a Decorator.

````python
decorated_greeting = uppercase_decorator(greeting)
print(decorated_greeting())
````

In the example above, `uppercase_decorator()` receives the `greeting()` function as its argument and returns a new function called `wrapper()`. When `wrapper()` is executed, it first calls `greeting()`, then converts the returned text to uppercase using the `upper()` method.

This approach allows us to extend the behavior of a function without modifying its original implementation.

>**In short:** A decorator is a Higher-Order Function designed specifically to wrap and enhance existing functions.

## 4.1 Using the @ Symbol

Python provides a more concise way to apply Decorators by using the `@` symbol.

Instead of writing:

````python
decorated_greeting = uppercase_decorator(greeting)
print(decorated_greeting())
````

we can place the Decorator directly above the function definition:

````python
@uppercase_decorator
def greeting():
    return "Welcome to Functional Programming"

print(greeting())
````

Both approaches produce exactly the same result. When Python encounters the `@uppercase_decorator` syntax, it automatically performs the following assignment behind the scenes:

**greeting = uppercase_decorator(greeting)**

Therefore, the `@` syntax is simply a cleaner and more readable way to apply a Decorator

## 4.2 Multiple Decorators

A function can also have more than one Decorator. In this case, each Decorator wraps the function one layer at a time. The Decorator closest to the function is executed first, followed by the Decorator above it.

Consider the following example.

````python
# Decorator 1: Add 10 to the function result
def add_ten(func):
    def wrapper(x):
        result = func(x)
        return result + 10
    return wrapper

# Decorator 2: Square the function result
def square_result(func):
    def wrapper(x):
        result = func(x)
        return result ** 2
    return wrapper
````

````python
# Decorator implementation
@square_result
@add_ten
def number(x):
    return x * 2

print(number(5))
````

Execution Flow:
1. The `number(5)` function is called.
2. The original function computes `5 × 2` = 10.
3. The `add_ten` Decorator is executed first, producing `10 + 10` = 20.
4. The result (`20`) is then passed to the `square_result` Decorator.
5. Finally, `20²` = 400 is returned.

Therefore, the final output is 400.

This execution order occurs because Decorators are applied from the bottom up. The Decorator closest to the function wraps the original function first, and each subsequent Decorator wraps the result of the previous one.

**Why Are Decorators Useful?**

Decorators provide a clean way to extend the behavior of existing functions without modifying their original source code. This helps keep the core logic of a function separate from additional functionality, making programs more modular, reusable, and easier to maintain.
