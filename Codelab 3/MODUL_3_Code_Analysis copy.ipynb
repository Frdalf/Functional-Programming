## Persiapan (fungsi dari bagian 1.1 – 1.4 & data)

```python
# Farid Al Farizi | NIM: 202410370110017
from functools import reduce

def is_positive(n):
    return n > 0

def add(a, b):
    return a + b

def divide(a, b):
    return a / b if b != 0 else "Error: division by zero"

def operate(func, x, y):
    return func(x, y)

def make_power(n):
    def power(x):
        return x ** n
    return power

operations = [add, divide, make_power, make_power(2)]

numbers = [1, 2, 3, 4, 5]
scores = [55, 80, 72, 65, 90, 45, 88, 60, 90, 50, 73]
student_scores = [[80, 75, 90], [60, 70, 65], [85, 88, 92]]
```

## Exercise 1: Reference or Result?

```python
# Farid Al Farizi | NIM: 202410370110017
check_number1 = is_positive      # referensi fungsi (tanpa kurung)
check_number2 = is_positive(1)   # hasil eksekusi -> True (dengan kurung)
```

```python
# Farid Al Farizi | NIM: 202410370110017
print(check_number1(-5))
try:
    print(check_number2(-5))
except TypeError as e:
    print("TypeError:", e)
```

## Exercise 2: The Parentheses Trap

```python
# Farid Al Farizi | NIM: 202410370110017
mixed = [
    divide,                # tanpa kurung -> fungsi
    divide(10, 5),         # dieksekusi -> 15
    make_power(2),      # fungsi hasil closure
    make_power(2)(4)    # dieksekusi -> 16
]

print(mixed)
```

```python
# Farid Al Farizi | NIM: 202410370110017
# Now try calling every element of the list by index
print("mixed[0](10, 5) =", mixed[0](10, 5))
print("mixed[2](4)     =", mixed[2](4))

for i in (1, 3):
    try:
        print(f"mixed[{i}](...)", mixed[i](4))
    except TypeError as e:
        print(f"mixed[{i}] ->", "TypeError:", e, "| nilainya:", mixed[i])
```

## Exercise 3: The Swappable Engine

```python
# Farid Al Farizi | NIM: 202410370110017
# Write your own subtraction function,
def subtract1(a, b):
    return a - b

# add it to the operations list
operations.append(subtract1)

# and do the execution
print("subtract =", operate(operations[4], 10, 5))
```

## Exercise 4: The Truthy Trap

```python
# Farid Al Farizi | NIM: 202410370110017
def is_even(x):
    return not x % 2 == 0

def is_odd_value(x):
    return not x % 2

print(list(filter(is_even, numbers)))
print(list(filter(is_odd_value, numbers)))
```

## Exercise 5: The Empty Shelf (NIM ganjil → tuple)

```python
# Farid Al Farizi | NIM: 202410370110017
products = [
    {"name": "Laptop", "stock": 3},
    {"name": "Mouse", "stock": 0},
    {"name": "Keyboard", "stock": 8},
    {"name": "Monitor", "stock": 0},
    {"name": "Flashdisk", "stock": 10}
]

def is_out_of_stock(product):
    """Return True when this product's stock is 0."""
    return product["stock"] < 5  # condition

out_of_stock = filter(is_out_of_stock, products)
print(tuple(out_of_stock))
```

```python
# Farid Al Farizi | NIM: 202410370110017
for product in out_of_stock:
    print(product)
# tidak ada output: filter object sudah habis dikonsumsi oleh tuple() di cell sebelumnya
```

## Exercise 6: The Vanishing Iterator

```python
# Farid Al Farizi | NIM: 202410370110017
def to_grade(score):
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    else:
        return "C"

grades = map(to_grade, scores)

print("Scores with Grades:")
# iterate to display the grades data using next()
for score in grades:
    print(score, '=', next(grades))

# Hand-trace:
# score=55 -> C | 80 -> B | 72 -> B | 65 -> C | 90 -> A | 45 -> C
# 88 -> A | 60 -> C | 90 -> A | 50 -> C | 73 -> B
```

## Exercise 7: Mapping Over Rows

```python
# Farid Al Farizi | NIM: 202410370110017
def bonus(score):
    return score + 5

try:
    print(list(map(bonus, to_grade(scores))))
except TypeError as e:
    print("TypeError:", e)
```

```python
# Farid Al Farizi | NIM: 202410370110017
# Apply map to each row (each student's scores)
bonus_scores = [list(map(bonus, row)) for row in student_scores]

print("original scores\t\t:", student_scores)
print("Total scores + bonus\t:", bonus_scores)
```

## Exercise 8: Flatten & Sum

```python
# Farid Al Farizi | NIM: 202410370110017
# Exercise 8a
reduce1 = reduce(add, student_scores)
print(reduce1)   # list datar (penggabungan), bukan satu angka
```

```python
# Farid Al Farizi | NIM: 202410370110017
# Exercise 8b
def row_total(row):
    return reduce(add, row)

row_totals = list(map(row_total, student_scores))   # total tiap baris
grand_total = reduce(add, row_totals)               # total keseluruhan

print("Row totals :", row_totals)
print("Grand total:", grand_total)
```

## Exercise 9: Think Recursively

```python
# Farid Al Farizi | NIM: 202410370110017
def sum_list(data):
    # base case: empty list -> 0
    if data == []:
        return 0
    # recursive case: first element/head + sum_list(the rest/tail)
    return data[0] + sum_list(data[1:])

flat_scores = reduce(add, student_scores)   # student_scores yang sudah di-flatten

print("sum_list :", sum_list(flat_scores))
print("reduce   :", grand_total)
print("Sama?    :", sum_list(flat_scores) == grand_total)
```
