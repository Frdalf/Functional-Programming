```python
# Nama: Farid Al Farizi
# NIM: 202410370110017

def total_sum(n):
    print(f"Current n: {n}")
    if n == 1:
        return 1
    else:
        return n + total_sum(n-1)

# Running the function
result = total_sum(3)
print("Final Result:", result)
```
